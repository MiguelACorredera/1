```python
# Install dependencies
import sys
!{sys.executable} -m pip install -q evaluate
!{sys.executable} -m pip install -q transformers

from datasets import load_dataset
import evaluate
from transformers import AutoTokenizer, AutoModelForSequenceClassification, TrainingArguments, Trainer
import numpy as np
import torch

# ===============================
# 1. Load Dataset
# ===============================
dataset = load_dataset(
    "csv",
    data_files="/content/sample_data/cleaned_liar_dataset.csv"
)

dataset = dataset["train"].train_test_split(test_size=0.2, seed=42)

# ===============================
# 2. Build rich input text
# ===============================
def build_text(example):
    parts = [
        example.get("statement", ""),
        example.get("subject", ""),
        example.get("speaker", ""),
        example.get("party_affiliation", ""),
        example.get("context", "")
    ]
    example["text"] = " [SEP] ".join([p for p in parts if p])
    return example

dataset = dataset.map(build_text)

# ===============================
# 3. Encode labels
# ===============================
unique_labels = sorted(dataset["train"].unique("label"))
label2id = {label: i for i, label in enumerate(unique_labels)}
id2label = {i: label for label, i in label2id.items()}

def encode_labels(example):
    example["label"] = label2id[example["label"]]
    return example

dataset = dataset.map(encode_labels)

print("Label mapping:", label2id)

# ===============================
# 4. Tokenize
# ===============================
tokenizer = AutoTokenizer.from_pretrained("microsoft/deberta-v3-base")

def tokenize_function(examples):
    return tokenizer(
        examples["text"],
        padding="max_length",
        truncation=True,
        max_length=384   # DeBERTa benefits from longer context
    )

tokenized_datasets = dataset.map(tokenize_function, batched=True)

# ===============================
# 5. Load DeBERTa-v3-base
# ===============================
model = AutoModelForSequenceClassification.from_pretrained(
    "microsoft/deberta-v3-base",
    num_labels=len(unique_labels),
    id2label=id2label,
    label2id=label2id
)

# ===============================
# 6. Training Arguments (older HF compatible)
# ===============================
training_args = TrainingArguments(
    output_dir="./results",
    do_train=True,
    do_eval=True,
    per_device_train_batch_size=8,     # DeBERTa is heavier than DistilBERT
    per_device_eval_batch_size=8,
    learning_rate=2e-5,
    num_train_epochs=5,                # Better convergence
    logging_steps=50,
    fp16=torch.cuda.is_available(),    # Use fp16 if GPU exists
    save_total_limit=1                 # Prevent disk overflow
)

# ===============================
# 7. Metrics
# ===============================
metric = evaluate.load("f1")

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=-1)
    return metric.compute(predictions=preds, references=labels, average="weighted")

# ===============================
# 8. Trainer
# ===============================
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets["train"],
    eval_dataset=tokenized_datasets["test"],
    compute_metrics=compute_metrics,
    tokenizer=tokenizer
)

# ===============================
# 9. Train
# ===============================
print("Starting DeBERTa training...")
trainer.train()

# ===============================
# 10. Evaluate
# ===============================
print("\nEvaluating model...")
results = trainer.evaluate()
print("\nFinal Results:")
print(results)

# ===============================
# 11. Save model
# ===============================
trainer.save_model("./deberta_liar_model")
tokenizer.save_pretrained("./deberta_liar_model")

print("\nModel saved to ./deberta_liar_model")
```
